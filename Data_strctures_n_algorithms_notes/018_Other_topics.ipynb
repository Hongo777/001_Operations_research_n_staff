```python
def modexp(x, n, p):
    # BASE CASE: Anything to the power of 0 is 1. We stop here.
    if n == 0:
        return 1
    
    # 1. Square the base and shrink it using modulo immediately
    t = (x * x) % p
    
    # 2. RECURSION: Cut the exponent in half (n // 2) and repeat the process
    tmp = modexp(t, n // 2, p)
    
    # 3. ODD CHECK: If the exponent was odd, we have one extra 'x' to multiply back in
    if n % 2 != 0:
        tmp = (tmp * x) % p
        
    return tmp
```

```python
def modexp(x, n, p):
    """
    Computes (x^n) % p efficiently using the divide-and-conquer approach.
    """
    # Base Case: x^0 is always 1
    if n == 0:
        return 1
    
    # Square the base and shrink it with modulo immediately
    t = (x * x) % p
    
    # Halve the exponent and solve the smaller problem recursively
    tmp = modexp(t, n // 2, p)
    
    # If the exponent was odd, multiply by the original base one more time
    if n % 2 != 0:
        tmp = (tmp * x) % p
        
    return tmp

# --- Example Usage ---
# Let's find 2^5 % 3
result = modexp(x=2, n=5, p=3)
print(f"The result of 2^5 mod 3 is: {result}") 

# Let's tackle the massive problem from your text: (31^254906) % 10
# (This finds the last digit of that giant number instantly!)
last_digit = modexp(x=31, n=254906, p=10)
print(f"The last digit of 31^254,906 is: {last_digit}")
```

```python
def gcd(a, b):
    if b == 0:
        return a
    elif a < b:
        return gcd(b, a)
    return gcd(a - b, b) # is really the same as the remainder of 𝑎 divided by 𝑏
print(gcd(77,7))

def gcd2(a, b):
    if b == 0:
        return a
    return gcd(b, a % b)
gcd2(8,2)
```

```python
def ext_gcd(x, y):
    """
    Returns a tuple (d, a, b) where:
    d = gcd(x, y)
    d = a*x + b*y
    """
    # Base case
    if y == 0:
        return (x, 1, 0)
    else:
        # Recursively find the values for the smaller remainder problem
        (d, a, b) = ext_gcd(y, x % y)
        # Update the coefficients using the formula from the text
        return (d, b, a - (x // y) * b)

def mod_inverse(x, m):
    """
    Finds the multiplicative inverse of x modulo m.
    It returns a number 'b' such that (x * b) % m == 1.
    """
    d, a, b = ext_gcd(m, x)
    if d != 1:
        raise ValueError(f"Inverse doesn't exist because gcd({x}, {m}) is {d}, not 1.")
    
    # 'b' might be negative, so we add m and mod by m to keep it positive
    return b % m

# --- Example Usage ---
# Let's use the text's example: x = 25, y = 9
gcd_val, a, b = ext_gcd(25, 9)
print(f"Extended GCD results for (25, 9):")
print(f"GCD = {gcd_val}, a = {a}, b = {b}")
print(f"Check the math: ({a} * 25) + ({b} * 9) = {a*25 + b*9}")

# Let's find the inverse of 3 modulo 7 (from the text's example)
# We want a number where (3 * inverse) % 7 == 1
inverse_3_mod_7 = mod_inverse(3, 7)
print(f"\nThe modular inverse of 3 modulo 7 is: {inverse_3_mod_7}")
print(f"Check the math: (3 * {inverse_3_mod_7}) % 7 = {(3 * inverse_3_mod_7) % 7}")
```
