## Queue

```python
# Completed implementation of a queue ADT
class Queue:
    def __init__(self):
        self.items = []
    def is_empty(self):
        return self.items == []
    def enqueue(self, item):
        self.items.insert(0,item) #this an O(n) operation , can we change the ends according to operation
    def dequeue(self):
        return self.items.pop()
    def size(self):
        return len(self.items)
    def head(self):
        return self. items[:5]
```

```python
q = Queue()
q.enqueue('hello')
q.enqueue('dog')
q.enqueue(3)
print( q.dequeue() )
q.head()
```

### hot potatoe problem

```python
#search more on the hot potatoe problem
def hot_potato(name_list, num):
    sim_queue = Queue()
    for name in name_list:
        sim_queue.enqueue(name)
    while sim_queue.size() > 1:
        for i in range(num): #do it like num times??
            sim_queue.enqueue(sim_queue.dequeue())
            #print(sim_queue.head())
        sim_queue.dequeue()
        
    return sim_queue.dequeue()


print(hot_potato(["Bill", "David", "Susan", "Jane", "Kent","Brad"], 7))     
```

```python
import random
class Printer:
    def __init__(self, ppm):
        self.page_rate = ppm
        self.current_task = None
        self.time_remaining = 0

    def tick(self):
        if self.current_task != None:
            self.time_remaining = self.time_remaining- 1 #does it mean we tick every second??
            if self.time_remaining <= 0:
                self.current_task = None
                
    def busy(self):
        if self.current_task != None:
            return True
        else:
            return False

    def start_next(self,new_task):
        self.current_task = new_task
        self.time_remaining = new_task.get_pages() * 60 / self.page_rate


class Task:
    def __init__(self, time):
        self.timestamp = time # arrival time
        self.pages = random.randrange(1, 21)
    def get_stamp(self):
        return self.timestamp
    def get_pages(self):
        return self.pages
    def wait_time(self, current_time):
        return current_time- self.timestamp

def simulation(num_seconds, pages_per_minute): #num_seconds is the time range we want to simulate the queue
    lab_printer = Printer(pages_per_minute)
    print_queue = Queue()
    waiting_times = [] #collects the waiting times
    
    for current_second in range(num_seconds):
        if new_print_task(): #this function is defined below
            task = Task(current_second)
            print_queue.enqueue(task) #add a task to the que
        if (not lab_printer.busy()) and (not print_queue.is_empty()):
            next_task = print_queue.dequeue() # note that all tasks were assighned a task class?...
            waiting_times.append(next_task.wait_time(current_second)) # but it was created with another 'current second'
            lab_printer.start_next(next_task) 
        lab_printer.tick()
    average_wait = sum(waiting_times) / len(waiting_times) #this the avage waiting times durin
    print("Average Wait %6.2f secs %3d tasks remaining."%(average_wait, print_queue.size())) 
##check? can't the size of the queue give the information of the avarage waiting time?
#and that the time task remaining is the size of the que after the time, num_seconds



def new_print_task(): #to determine if a new task is present in a second...its a second in this case
    num = random.randrange(1, 181)
    if num == 180:
        return True
    else:
        return False

#implemenataion 
for i in range(10): #we will run 10 independent trials..
    #we will run the simulation for a period of 60 minutes (3,600 seconds) using a pagerate of five pages per minute. 
    simulation(3600, 5)
```
