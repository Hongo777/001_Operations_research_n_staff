```python
# entry A_in
A_in={ "food_1":{"collories":400,
                  "Chocolate":3,
                  "Sugar":2,
                  "Fat":2
                },
        "food_2":{"collories":200,
                  "Chocolate":2,
                  "Sugar":2,
                  "Fat":4
                },
        "food_3":{"collories":150,
                  "Sugar":4,
                  "Fat":1
                },
        "food_4":{"collories":500,
                  "Sugar":4,
                  "Fat":5
                }
     }
                
    #finish of A_in

#split them
A_id=list(A_in.keys())
A=list(A_in.values())

import numpy as np

A_raws=[]
Fields=[]



#create fields
for i in range(len(A)):
    for i2 in range(len(A[i])):
        if list(A[i].keys())[i2] in Fields:
            pass
        else:
            Fields.append(list(A[i].keys())[i2])
#creat A
for i in range(len(A)):
    Ar=[0]*(len(Fields))
    for i2 in range(len(A[i])):
        #from here we have value and key of corresponding to i2, the get its index in fields then(if) with i2th key, update Ar[if] with the i2th value
        Ar[Fields.index(list(A[i].keys())[i2])]=list(A[i].values())[i2]
        #Ar[i2]=i2+1
    A_raws.append(Ar)
Matrix_A_T=np.array(A_raws)
Matrix_A=Matrix_A_T.T

# getting vector b which is this for this example:[500,6,10,8]

b=[]
for i in range(len(Fields)):
    b0=input(f"enter constarint for {Fields[i]} ,our example is [500,6,10,8]:")
    b.append(int(b0))
#getting C which is this for this case:[50,20,30,80]
print(".......")
C=[]
for i in range(len(A_id)):
    C0=input(f"enter Cost for {A_id[i]} ,our example is [50,20,30,80]:")
    C.append(int(C0))
    
print("  ")
print(f"your constraint Fields are: {Fields}:")
print(f"your vector b constarints are: {b}:")
print(f"your decision varaibles are: {A_id}:")
print(f"your Cost matrix is: {C}:")
print("your matrix A is :")
print(Matrix_A)

print(".......")

#solving staff
from gekko import GEKKO
m = GEKKO(remote=False)

b=np.array(b)
C=np.array(C)
m = GEKKO(remote=False)
x = m.Array(m.Var,len(A_id),lb=0)
m.Minimize(C@x)
Constraint_equations = Matrix_A@x
m.Equations([Constraint_equations[i]>=b[i] for i in range(len(b))])
m.options.SOLVER=1 #to make sure its integer
m.solve(disp=False)

# Print the results
print("Objective function: ", m.options.objfcnval)
print("Optimal solution: ", x)
print("that is")
for i in range(len(A_id)):
    print(f'{A_id[i]}:',x[i].value[0])
print(f"\nMinimum Daily Cost: ${m.options.OBJFCNVAL:.2f}")
```

```python
#best best Code code
SP=np.array([70,60,50])
D=np.array([3000,2000,1000])
Ad=np.array([10,10,10])
GEPi=np.array([[10,8,6]])
BEPi=np.array([[0.01,0.02,0.01]])

#Resources
BP=np.array([45,35,25])
MP=np.array([5000,5000,5000])
GEPRi=np.array([[12,6,8]])
BEPRi=np.array([[0.005,0.02,0.03]])

RPC=4
maxP=14000

p=len(SP)
r=len(BP)

from gekko import GEKKO
m = GEKKO(remote=False)

x = m.Array(m.Var,(r,p),lb=0)
a = m.Array(m.Var,p,lb=0)
xs=[]
cost=[]


for i in range(r):
    xs.append(m.Intermediate(m.sum(x[i])))
    cost.append(m.Intermediate(BP[i]*xs[i] ))
    m.Equations( [ xs[i]<=MP[i]  ])
m.Equations( [ m.sum(xs)<= maxP  ])


revenue=[]
#z=x.T
z=[]
zs=[]


for j in range(p):
    zc=[]
    for i in range(r):
        zc.append(x[i][j])
    z.append(zc)
    zs.append(m.Intermediate(m.sum(zc) ))
    revenue.append(m.Intermediate( SP[j]*zs[j]))
    m.Equations( [ zs[j]==D[j]+ Ad[j]*a[j] ])

    for i in range(len(GEPi)):# Good elements
        #since matric multiplication is just element multiplication then addition
        m.Equations( [ m.sum( GEPRi[i]*z[j])>=GEPi[i][j]*zs[j]   ])
    #bad elements
    for i in range(len(BEPi)):
        m.Equations( [ m.sum( BEPRi[i]*z[j])<=BEPi[i][j]*zs[j]   ])

Ad_cost=m.Intermediate(m.sum(a))
costT=m.Intermediate(m.sum(cost))
tatal_revenue=m.Intermediate(m.sum(revenue))
production=m.Intermediate(m.sum(xs)*RPC)
profit=m.Intermediate(tatal_revenue-Ad_cost-production-costT)

m.Maximize(profit)
m.options.SOLVER=1
m.solve(disp=False)
print('profit= ' + str(profit.value[0]))
x,a
```
