```python
# entry A_in
A_in={ "Brown":{"collories":400,
                  "oz":3,
                  "sugar":2,
                  "fat":2
                },
        "Choco":{"collories":200,
                  "oz":2,
                  "sugar":2,
                  "fat":4
                },
        "Cola":{"collories":150,
                  "sugar":4,
                  "fat":1
                },
        "Pine":{"collories":500,
                  "sugar":4,
                  "fat":5
                }
     }#finish of A_in

#split them
A_id=list(A_in.keys())
A=list(A_in.values())

import numpy as np

A_raws=[]
Fields=[]



#create fields
for i in range(len(A)):
    for i2 in range(len(A[i])):
        if list(A[i].keys())[i2] in Fields:
            pass
        else:
            Fields.append(list(A[i].keys())[i2])
#creat A
for i in range(len(A)):
    Ar=[0]*(len(Fields))
    for i2 in range(len(A[i])):
        #from here we have value and key of corresponding to i2, the get its index in fields then(if) with i2th key, update Ar[if] with the i2th value
        Ar[Fields.index(list(A[i].keys())[i2])]=list(A[i].values())[i2]
        #Ar[i2]=i2+1
    A_raws.append(Ar)
Matrix_A_T=np.array(A_raws)
Matrix_A=Matrix_A_T.T

# getting vector b which is this for this example:[500,6,10,8]
b=[]
for i in range(len(Fields)):
    b0=input(f"enter constarint for {Fields[i]}:")
    b.append(int(b0))
#getting C which is this for this case:[50,20,30,80]
print(".......")
C=[]
for i in range(len(A_id)):
    C0=input(f"enter Cost for {A_id[i]}:")
    C.append(int(C0))
    
print("  ")
print(f"your constraint Fields are: {Fields}:")
print(f"your vector b constarints are: {b}:")
print(f"your decision varaibles are: {A_id}:")
print(f"your Cost matrix is: {C}:")
print("your matrix A is :")
print(Matrix_A)
```

```python
C
```

```python
from gekko import GEKKO
m = GEKKO(remote=False)

b=np.array(b)
C=np.array(C)
m = GEKKO(remote=False)
x = m.Array(m.Var,len(A_id),lb=0)
m.Minimize(C@x)
Constraint_equations = Matrix_A@x
m.Equations([Constraint_equations[i]>=b[i] for i in range(len(b))])
m.options.SOLVER=1 #to make sure its integer
m.solve(disp=False)

# Print the results
print("Objective function: ", m.options.objfcnval)
print("Optimal solution: ", x)
```

```python
from scipy.optimize import linprog

# Define the objective function
c = [-1, 4] # minimize -x + 4y

# Define the constraints
A_u = np.array([[3, 2], [1, 5]])
b_u = np.array([10,10])

# Define the bounds
b = (0, 10) # 0 <= x <= 10

# Solve
res = linprog(c, A_ub=A_u, b_ub=b_u, bounds=(b,b),
              integrality=[1,1])

# Print the results
print("Objective function: ", res.fun)
print("Optimal solution: ", res.x)
```
